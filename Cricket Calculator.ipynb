```python
# Chase Master Cricket Calculator

target = int(input("Enter target score: "))
current = int(input("Enter current score: "))
overs = input("Enter overs bowled (e.g. 37.4): ")
wickets_lost = int(input("Enter wickets lost: "))

over_parts = overs.split(".")
complete_overs = int(over_parts[0])
balls_in_over = int(over_parts[1])

balls_bowled = complete_overs * 6 + balls_in_over

total_balls = 300
balls_remaining = total_balls - balls_bowled

overs_remaining = f"{balls_remaining // 6}.{balls_remaining % 6}"

runs_required = target - current

if balls_bowled > 0:
    crr = current / (balls_bowled / 6)
else:
    crr = 0


if balls_remaining > 0:
    rrr = max(0, runs_required) / (balls_remaining / 6)
else:
    rrr = 0 if runs_required <= 0 else float('inf')
wickets_remaining = 10 - wickets_lost

chase_feasible = (
    runs_required <= balls_remaining * 6
    and balls_remaining >= 0
    and wickets_remaining > 0
)

chase_comfortable = (
    rrr <= crr and wickets_lost < 7
)

print("\n------ CHASE MASTER RESULTS ------")

print("Balls bowled:", balls_bowled)
print("Balls remaining:", balls_remaining)
print("Overs remaining:", overs_remaining)
print("Runs required:", runs_required)
print("Current run rate:", round(crr, 2))
print("Required run rate:", round(rrr, 2))
print("Wickets remaining:", wickets_remaining)
print("Chase feasible:", chase_feasible)
print("Chase comfortable:", chase_comfortable)
```
