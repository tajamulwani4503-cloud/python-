TYPES OF OPERATORS IN PYTHON
1. ARTHEMETIC OPERATOR { +, -, *, /, % }
2. RELATIONAL OPERATOR { <, <=, >, >=, ==, != }
3. LOGICAL OPERATOR { AND, OR, NOT }
4. BITWISE OPERATOR { &, |, ^, ~, <<, >> }
5. ASSIGNMENT OPERATOR { =, +=, -=, *=, %= }
6. TERNARY OPERATOR { X IF CINDITION ELSE Y }
7. IDENTITY OPERATIOR { IS, IS NOT }

```python
# Arthemetic operator 
a = 20
b = 15 

print("Addition:", a + b)

print("Subtraction:", a - b)

print("Multiplication:", a * b)

print("Division:", a / b)

print("Floor Division:", a // b)

print("Modulus:", a % b)

print("Exponentiation:", a ** b)


```

```python
# Comparison operator
a = 10
b = 22

print("Equal:", a == b)

print("Not Equal:", a != b)

print("Less Than:", a < b)

print("Greater Than:", a > b)

print("Less Than or Equal:", a <= b)

print("Greater Than or Equal:", a >= b)
```

```python
# Logical operator

 a = True
 b = False
 print( a  and b)

 print( a or b)

 print( not a)
```

```python
# Bitwise operator 
a = 10
b = 4

print("Bitwise AND:", a & b)

print("Bitwise OR:", a | b)

print("Bitwise XOR:", a ^ b)

print("Bitwise NOT:", ~a)

print("Left Shift:", a << 2)

print("Right Shift:", a >> 2)
```

```python
# Assignment operator
a = 20 
b = 10 

print(b)
b += a

print(b)
b -= a

print(b)
b *= a

print(b)
b <<= a

print(b)
b >>= a

print(b)
b /= a

print(b)
b %= a

print(b)
b **= a
```

```python
# Identity operator 
a = 10
b = 20 
c = a

print(a is not b)

print(a is c)
```

```python
# Membership operator 
a = 24
b = 20 
my_list = [10, 20, 30, 40, 50]

if (a not in my_list):
    print("a is not preasent in my_list")
else:
    print("a is in my_list")

if (b  in my_list):
    print("b is present in my list")
else:
    print("b is not present in my list")
```

```python
# Ternary operator
a = 10 
b = 20 

min = a if a < b else b

print(min)
```
